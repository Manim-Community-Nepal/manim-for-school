```python
from manim import *

class FullCongruencyLesson(Scene):
    def construct(self):
        # फन्ट सेटिङ (तपाईंको सिस्टममा भएको फन्ट राख्नुहोला)
        nepali_font = "Noto Sans Devanagari"

        # ==========================================
        # PART 1: Introduction to Congruency
        # ==========================================
        part1_title = Text("Congruent Triangles (अनुरूप त्रिभुजहरू)", font=nepali_font, font_size=40, color=BLUE)
        part1_subtitle = Text("Part - 1: Concept", font=nepali_font, font_size=32, color=YELLOW)
        title_group1 = VGroup(part1_title, part1_subtitle).arrange(DOWN).to_edge(UP)
        
        self.play(Write(title_group1))
        
        def_text = Text(
            "आकार (Shape) र नाप (Size) ठ्याक्कै बराबर भएका आकृतिहरू", 
            font=nepali_font, font_size=28
        ).next_to(title_group1, DOWN, buff=0.5)
        
        self.play(FadeIn(def_text, shift=UP))
        self.wait(1)

        # Part 1 Triangles
        t1_vertices = [np.array([-4, -2.5, 0]), np.array([-1, -2.5, 0]), np.array([-3, 0.5, 0])]
        tri1 = Polygon(*t1_vertices, color=WHITE, fill_opacity=0.3, fill_color=BLUE)
        
        tri2 = tri1.copy().set_color(WHITE).set_fill(color=GREEN, opacity=0.5)
        tri2.shift(RIGHT * 5)

        self.play(DrawBorderThenFill(tri1), DrawBorderThenFill(tri2))
        self.wait(1)

        overlap_text = Text("यिनीहरू एक-अर्कामा पूर्ण रूपमा खप्टिन्छन् (Overlap)", font=nepali_font, font_size=26, color=YELLOW)
        overlap_text.move_to(UP * 0.5) 
        self.play(Write(overlap_text))

        # Overlap Animation
        self.play(tri2.animate.move_to(tri1.get_center()), run_time=2)
        self.play(Indicate(tri2, color=YELLOW, scale_factor=1.1))
        self.wait(1)
        
        # Move back and show Congruence Symbol
        self.play(tri2.animate.shift(RIGHT * 5), run_time=1.5)
        
        cong_symbol = MathTex(r"\cong", font_size=72, color=YELLOW).move_to(DOWN * 1)
        self.play(Write(cong_symbol))
        self.play(Wiggle(cong_symbol))
        self.wait(2)

        # Clear screen completely for Part 2
        self.clear()


        # ==========================================
        # PART 2: Conditions of Congruency
        # ==========================================
        part2_title = Text("Conditions of Congruent Triangles", font=nepali_font, font_size=40, color=BLUE)
        part2_subtitle = Text("Part - 2", font=nepali_font, font_size=32, color=YELLOW)
        title_group2 = VGroup(part2_title, part2_subtitle).arrange(DOWN).to_edge(UP)
        
        self.play(Write(title_group2))

        conditions = [
            ("१. SSS – तीनै भुजा बराबर", "SSS"),
            ("२. SAS – दुई भुजा र बीचको कोण बराबर", "SAS"),
            ("३. ASA – दुई कोण र बीचको भुजा बराबर", "ASA"),
            ("४. AAS – दुई कोण र एउटा भुजा बराबर", "AAS"),
            ("५. RHS – Right angle, Hypotenuse र एउटा भुजा", "RHS")
        ]

        current_text = Text("", font=nepali_font)

        for text_str, cond_type in conditions:
            # Condition Text Update
            new_text = Text(text_str, font=nepali_font, font_size=30, color=WHITE)
            new_text.next_to(title_group2, DOWN, buff=0.5)
            self.play(ReplacementTransform(current_text, new_text))
            current_text = new_text

            # Triangle Coordinates
            if cond_type == "RHS":
                # Right Angled Triangles for RHS
                vA, vB, vC = np.array([-4, 1, 0]), np.array([-4, -2.5, 0]), np.array([-1.5, -2.5, 0])
                vP, vQ, vR = np.array([2, 1, 0]), np.array([2, -2.5, 0]), np.array([4.5, -2.5, 0])
            else:
                # Normal Triangles for others
                vA, vB, vC = np.array([-3, 1, 0]), np.array([-4.5, -2.5, 0]), np.array([-1.5, -2.5, 0])
                vP, vQ, vR = np.array([3, 1, 0]), np.array([1.5, -2.5, 0]), np.array([4.5, -2.5, 0])

            # Recreating Triangles for every property
            t1_base, t1_right, t1_left = Line(vB, vC), Line(vC, vA), Line(vA, vB)
            t2_base, t2_right, t2_left = Line(vQ, vR), Line(vR, vP), Line(vP, vQ)
            
            t1 = VGroup(t1_base, t1_right, t1_left)
            t2 = VGroup(t2_base, t2_right, t2_left)

            # Labels
            label_A = MathTex("A").next_to(vA, UP, buff=0.1)
            label_B = MathTex("B").next_to(vB, LEFT, buff=0.1)
            label_C = MathTex("C").next_to(vC, RIGHT, buff=0.1)
            
            label_P = MathTex("P").next_to(vP, UP, buff=0.1)
            label_Q = MathTex("Q").next_to(vQ, LEFT, buff=0.1)
            label_R = MathTex("R").next_to(vR, RIGHT, buff=0.1)
            
            labels = VGroup(label_A, label_B, label_C, label_P, label_Q, label_R)

            self.play(Create(t1), Create(t2), run_time=1.5)
            self.play(Write(labels), run_time=1)
            
            # Highlighting Logic
            if cond_type == "SSS":
                self.highlight_sides(t1_base, t2_base, RED)
                self.highlight_sides(t1_right, t2_right, GREEN)
                self.highlight_sides(t1_left, t2_left, ORANGE)

            elif cond_type == "SAS":
                self.highlight_sides(t1_left, t2_left, RED)
                self.highlight_angles(vA, vB, vC, vP, vQ, vR, YELLOW) # Angle B & Q
                self.highlight_sides(t1_base, t2_base, GREEN)

            elif cond_type == "ASA":
                self.highlight_angles(vA, vB, vC, vP, vQ, vR, YELLOW) # Angle B & Q
                self.highlight_sides(t1_base, t2_base, RED)           # Base BC & QR
                self.highlight_angles(vB, vC, vA, vQ, vR, vP, ORANGE) # Angle C & R

            elif cond_type == "AAS":
                self.highlight_angles(vB, vA, vC, vQ, vP, vR, YELLOW) # Angle A & P
                self.highlight_angles(vA, vB, vC, vP, vQ, vR, ORANGE) # Angle B & Q
                self.highlight_sides(t1_right, t2_right, GREEN)       # Side AC & PR

            elif cond_type == "RHS":
                # Right Angle Marking
                ra1 = RightAngle(t1_left, t1_base, quadrant=(-1,1), color=YELLOW, length=0.4)
                ra2 = RightAngle(t2_left, t2_base, quadrant=(-1,1), color=YELLOW, length=0.4)
                self.play(Create(ra1), Create(ra2))
                
                self.highlight_sides(t1_right, t2_right, RED) # Hypotenuse
                self.highlight_sides(t1_base, t2_base, GREEN) # Base

            # Congruency Symbol
            cong_symbol2 = MathTex(r"\cong", font_size=60, color=YELLOW).move_to(DOWN * 0.75)
            self.play(GrowFromCenter(cong_symbol2))
            self.play(Wiggle(cong_symbol2))
            self.wait(1.5)
            
            # --- OVERLAP FIX ---
            # Remove EVERYTHING on screen except the titles and current text
            objects_to_keep = [title_group2, current_text]
            objects_to_remove = [mob for mob in self.mobjects if mob not in objects_to_keep]
            
            # Fade out old triangles, highlighted lines, angles, symbols, etc.
            self.play(*[FadeOut(mob) for mob in objects_to_remove])

        # Final Conclusion
        final_text = Text("यसरी यी ५ नियमहरूले त्रिभुजको Congruency प्रमाणित गर्छन्।", font=nepali_font, font_size=28, color=GREEN_C)
        final_text.next_to(title_group2, DOWN, buff=1.5)
        self.play(ReplacementTransform(current_text, final_text))
        self.wait(3)

    # ---------------------------------------------------------
    # Helper Functions
    # ---------------------------------------------------------
    def highlight_sides(self, s1, s2, color):
        h_s1 = s1.copy().set_color(color).set_stroke(width=10)
        h_s2 = s2.copy().set_color(color).set_stroke(width=10)
        
        self.play(Create(h_s1), Create(h_s2), run_time=0.8)
        self.play(Indicate(h_s1, color=color), Indicate(h_s2, color=color), run_time=0.8)
        self.add(h_s1, h_s2) # We keep them on screen, but they will be cleaned by the OVERLAP FIX

    def highlight_angles(self, p1_start, p1_vertex, p1_end, p2_start, p2_vertex, p2_end, color):
        a1_start_angle = Line(p1_vertex, p1_start).get_angle()
        a1_end_angle = Line(p1_vertex, p1_end).get_angle()
        a1_angle = (a1_end_angle - a1_start_angle) % TAU
        if a1_angle > PI: a1_angle -= TAU
        
        angle1 = Sector(radius=0.6, arc_center=p1_vertex, start_angle=a1_start_angle, angle=a1_angle, color=color, fill_opacity=0.6)
        
        a2_start_angle = Line(p2_vertex, p2_start).get_angle()
        a2_end_angle = Line(p2_vertex, p2_end).get_angle()
        a2_angle = (a2_end_angle - a2_start_angle) % TAU
        if a2_angle > PI: a2_angle -= TAU
        
        angle2 = Sector(radius=0.6, arc_center=p2_vertex, start_angle=a2_start_angle, angle=a2_angle, color=color, fill_opacity=0.6)
        
        self.play(DrawBorderThenFill(angle1), DrawBorderThenFill(angle2), run_time=0.8)
        self.play(Indicate(angle1, color=color), Indicate(angle2, color=color), run_time=0.6)

%manim -qk -v warning FullCongruencyLesson
```
