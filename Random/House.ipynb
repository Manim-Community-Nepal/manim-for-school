```python
from manim import *
import numpy as np

class Perfect3DHouse(ThreeDScene):
    def construct(self):
        # 1. Initial Camera Setup (Dead-on 2D view)
        # phi=90 puts the camera flat on the ground plane.
        # theta=-90 points it directly at the front of the house (Y = -1.25).
        self.set_camera_orientation(phi=90 * DEGREES, theta=-90 * DEGREES)

        # 2. Colors
        color_wall = "#F5F5DC"  # Beige
        color_roof = "#800000"  # Maroon
        color_trim = WHITE
        color_door = "#8B4513"  # Brown
        color_glass = BLUE_D

        # 3. Ground Plane (Z=0)
        ground = Polygon(
            [-10, -10, 0], [10, -10, 0], [10, 10, 0], [-10, 10, 0],
            fill_color="#2E8B57", fill_opacity=1, stroke_width=0
        )

        # 4. House Structure
        front_wall = Polygon([-1.5, -1.25, 0], [1.5, -1.25, 0], [1.5, -1.25, 2], [-1.5, -1.25, 2], fill_color=color_wall, fill_opacity=1, stroke_color=color_trim)
        back_wall = Polygon([-1.5, 1.25, 0], [1.5, 1.25, 0], [1.5, 1.25, 2], [-1.5, 1.25, 2], fill_color=color_wall, fill_opacity=1, stroke_color=color_trim)
        left_wall = Polygon([-1.5, -1.25, 0], [-1.5, 1.25, 0], [-1.5, 1.25, 2], [-1.5, -1.25, 2], fill_color=color_wall, fill_opacity=1, stroke_color=color_trim)
        right_wall = Polygon([1.5, -1.25, 0], [1.5, 1.25, 0], [1.5, 1.25, 2], [1.5, -1.25, 2], fill_color=color_wall, fill_opacity=1, stroke_color=color_trim)

        # 5. Roof Structure
        front_gable = Polygon([-1.5, -1.25, 2], [1.5, -1.25, 2], [0, -1.25, 3.5], fill_color=color_wall, fill_opacity=1, stroke_color=color_trim)
        back_gable = Polygon([-1.5, 1.25, 2], [1.5, 1.25, 2], [0, 1.25, 3.5], fill_color=color_wall, fill_opacity=1, stroke_color=color_trim)
        left_roof = Polygon([-1.5, -1.25, 2], [-1.5, 1.25, 2], [0, 1.25, 3.5], [0, -1.25, 3.5], fill_color=color_roof, fill_opacity=1, stroke_color=color_trim)
        right_roof = Polygon([1.5, -1.25, 2], [1.5, 1.25, 2], [0, 1.25, 3.5], [0, -1.25, 3.5], fill_color=color_roof, fill_opacity=1, stroke_color=color_trim)

        # 6. Details
        door = Polygon([-0.4, -1.26, 0], [0.4, -1.26, 0], [0.4, -1.26, 1.2], [-0.4, -1.26, 1.2], fill_color=color_door, fill_opacity=1, stroke_color=BLACK)
        
        def create_arched_window(x_offset):
            base = Rectangle(width=0.6, height=0.6).set_fill(color_glass, opacity=0.8).set_stroke(WHITE)
            arch = Sector(radius=0.3, angle=PI, start_angle=0).set_fill(color_glass, opacity=0.8).set_stroke(WHITE)
            arch.next_to(base, UP, buff=0)
            win = VGroup(base, arch)
            win.rotate(PI/2, axis=RIGHT)
            win.move_to(np.array([x_offset, -1.26, 1.2]))
            return win

        window_left = create_arched_window(-0.8)
        window_right = create_arched_window(0.8)

        # 7. Grouping
        house = VGroup(
            front_wall, back_wall, left_wall, right_wall,
            front_gable, back_gable, left_roof, right_roof,
            door, window_left, window_right
        )
        environment = VGroup(ground)

        # 8. Animation Sequence
        self.add(environment) # Ground is added quietly in the background
        
        # Step A: Construct the house while viewing strictly from the front
        self.play(Create(house), run_time=4)
        self.wait(1)

        # Step B: Smoothly tilt the camera up and to the side to reveal the 3D depth
        self.move_camera(phi=70 * DEGREES, theta=-45 * DEGREES, run_time=2)
        
        # Step C: Begin the continuous showcase rotation
        self.begin_ambient_camera_rotation(rate=0.15)
        self.wait(8)

%manim -ql -v warning Perfect3DHouse
```
