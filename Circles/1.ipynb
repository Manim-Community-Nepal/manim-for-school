```python
from manim import *

class CircleGeometry(Scene):
    def construct(self):
        # --- 1. Top Section: The Question ---
        question = Text(
            "Find the equation of a circle centered at (5, 6) and passing\n"
            "through the intersection of x + 2y = 6 and 3x - y = 4.",
            font_size=30, color=YELLOW
        ).to_edge(UP, buff=0.3)
        
        horiz_line = Line(LEFT, RIGHT, color=WHITE).scale(7).next_to(question, DOWN, buff=0.2)
        vert_line = Line(UP, DOWN, color=GRAY).scale(2.5).shift(LEFT * 1.0) # Shifted for more text space

        self.play(Write(question), Create(horiz_line), Create(vert_line))

        # --- 2. Right Section: The Graph ---
        axes = Axes(
            x_range=[0, 10, 1], y_range=[0, 10, 1],
            x_length=5, y_length=5,
            axis_config={"include_tip": True}
        ).shift(RIGHT * 3.5 + DOWN * 0.5)
        labels = axes.get_axis_labels(x_label="x", y_label="y")
        
        self.play(Create(axes), Write(labels))

        # --- 3. Left Section: Part A (Intersection) ---
        step1_title = Text("1. Solve for Intersection (x, y):", font_size=24, color=BLUE).to_edge(LEFT).shift(UP * 1.5)
        eqs = MathTex(
            r"x + 2y = 6 \quad \dots (i)\\",
            r"3x - y = 4 \implies y = 3x - 4 \quad \dots (ii)",
            font_size=22
        ).next_to(step1_title, DOWN, aligned_edge=LEFT)
        
        solve_steps = MathTex(
            r"\text{Sub (ii) in (i): } x + 2(3x-4) = 6\\",
            r"7x - 8 = 6 \implies 7x = 14 \implies x = 2\\",
            r"\text{From (ii): } y = 3(2) - 4 = 2\\",
            r"\therefore \text{Intersection (x, y) = (2, 2)}",
            font_size=22
        ).next_to(eqs, DOWN, aligned_edge=LEFT)

        line1 = axes.plot(lambda x: (6 - x) / 2, x_range=[0, 6], color=BLUE_B)
        line2 = axes.plot(lambda x: 3 * x - 4, x_range=[1.3, 4], color=GREEN_B)
        inter_point = Dot(axes.c2p(2, 2), color=RED)
        inter_label = MathTex("(2, 2)", font_size=20).next_to(inter_point, DL, buff=0.1)

        self.play(FadeIn(step1_title), Write(eqs))
        self.play(Create(line1), Create(line2))
        self.play(Write(solve_steps))
        self.play(Create(inter_point), Write(inter_label))
        self.wait(2)

        # --- 4. Clearing Part A to make room for Part B ---
        # This prevents the text from going to the bottom of the screen
        self.play(FadeOut(step1_title), FadeOut(eqs), FadeOut(solve_steps))

        # --- 5. Left Section: Part B (Circle Parameters & Equation) ---
        step2_title = Text("2. Equation of the Circle:", font_size=24, color=BLUE).to_edge(LEFT).shift(UP * 1.5)
        
        params = MathTex(
            r"\text{Center (h, k) = (5, 6)}\\",
            r"\text{Passes through (x, y) = (2, 2)}\\",
            r"\text{Let } r \text{ be the radius.}",
            font_size=22
        ).next_to(step2_title, DOWN, aligned_edge=LEFT)

        center_dot = Dot(axes.c2p(5, 6), color=YELLOW)
        center_label = MathTex("(5, 6)", font_size=20).next_to(center_dot, UR, buff=0.1)

        r_calc = MathTex(
            r"r = \sqrt{(x-h)^2 + (y-k)^2}\\",
            r"r = \sqrt{(2-5)^2 + (2-6)^2} = 5",
            font_size=22
        ).next_to(params, DOWN, aligned_edge=LEFT)

        final_eq = MathTex(
            r"\text{Now, } (x-h)^2 + (y-k)^2 = r^2\\",
            r"(x-5)^2 + (y-6)^2 = 5^2\\",
            r"x^2 + y^2 - 10x - 12y + 36 = 0",
            font_size=22, color=YELLOW
        ).next_to(r_calc, DOWN, aligned_edge=LEFT)

        radius_line = Line(axes.c2p(5, 6), axes.c2p(2, 2), color=WHITE, stroke_width=2)
        circle = Circle(radius=axes.get_x_unit_size() * 5, color=YELLOW).move_to(axes.c2p(5, 6))

        self.play(FadeIn(step2_title), Write(params))
        self.play(Create(center_dot), Write(center_label))
        self.play(Create(radius_line), Write(r_calc))
        self.wait(1)
        self.play(Write(final_eq), Create(circle))
        self.wait(5)

%manim -qk -v warning CircleGeometry
```
