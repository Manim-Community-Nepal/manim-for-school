```python
from manim import *

class NepaliMathProblem(ThreeDScene):
    def construct(self):
        # --- CONFIGURATION ---
        nep_font = "Noto Sans Devanagari" 
        r_val, h_cone = 0.7, 2.4 
        res = (10, 10) # Low res for speed

        # --- 1. INITIAL SCENE ---
        title = Text("ठोस वस्तुको विस्तृत व्याख्या", font=nep_font, color=BLUE).scale(0.7)
        self.add_fixed_in_frame_mobjects(title)
        title.to_edge(UP)

        cone = Cone(base_radius=r_val, height=h_cone, direction=UP, 
                    resolution=res, fill_opacity=0.4, color=GOLD)
        cone.shift(UP * h_cone / 2)
        
        hemi = Sphere(radius=r_val, v_range=[PI / 2, PI], u_range=[0, TAU],
                      resolution=res, fill_opacity=0.4, color=ORANGE)
        hemi.move_to(ORIGIN)
        
        composite = VGroup(cone, hemi)
        self.set_camera_orientation(phi=75 * DEGREES, theta=-45 * DEGREES)
        self.play(Create(composite))

        # --- 2. EXPLODED VIEW (The "Why") ---
        # Moving them apart to show they are two shapes
        self.play(
            cone.animate.shift(UP * 0.5),
            hemi.animate.shift(DOWN * 0.5),
            run_time=1.5
        )

        # Labels for the exploded parts
        cone_label = Text("शंकु (Cone)", font=nep_font, font_size=20, color=GOLD)
        hemi_label = Text("अर्धगोला (Hemisphere)", font=nep_font, font_size=20, color=ORANGE)
        
        self.add_fixed_in_frame_mobjects(cone_label, hemi_label)
        cone_label.move_to(RIGHT * 3 + UP * 1)
        hemi_label.move_to(RIGHT * 3 + DOWN * 1)

        self.play(Write(cone_label), Write(hemi_label))
        self.wait(1)

        # --- 3. THE CALCULATION OVERLAY ---
        # Shift everything to clear space
        self.play(
            FadeOut(cone_label), FadeOut(hemi_label),
            VGroup(cone, hemi).animate.scale(0.8).to_edge(LEFT, buff=1),
            run_time=1
        )

        # Step-by-step math
        step1 = MathTex(r"V_1 = \frac{1}{3}\pi r^2 h", color=GOLD, font_size=30)
        step2 = MathTex(r"V_2 = \frac{2}{3}\pi r^3", color=ORANGE, font_size=30)
        total_v = MathTex(r"V_{Total} = V_1 + V_2 = 1950.67 \text{ cm}^3", font_size=34)
        
        math_vgroup = VGroup(step1, step2, total_v).arrange(DOWN, aligned_edge=LEFT).to_edge(RIGHT, buff=1)
        self.add_fixed_in_frame_mobjects(math_vgroup)
        
        for step in math_vgroup:
            self.play(Write(step))
            self.wait(0.5)

        # --- 4. MELTING TRANSFORMATION ---
        # "Melt" the shapes back together then into a cylinder
        self.play(cone.animate.shift(DOWN * 0.5), hemi.animate.shift(UP * 0.5), run_time=0.5)
        
        melt_title = Text("पगालेर बेलना बनाउँदाको प्रक्रिया", font=nep_font, color=PINK, font_size=28).to_edge(UP)
        self.add_fixed_in_frame_mobjects(melt_title)
        
        # New Cylinder
        cyl_h = 1.267
        cylinder = Cylinder(radius=r_val, height=cyl_h, resolution=res, color=BLUE, fill_opacity=0.6)
        cylinder.move_to(cone.get_center() + DOWN*0.8)

        self.play(
            FadeOut(title),
            FadeIn(melt_title),
            ReplacementTransform(VGroup(cone, hemi), cylinder),
            math_vgroup.animate.scale(0.7).to_edge(UP).shift(LEFT*1),
            run_time=2
        )

        # --- 5. FINAL EXPLANATION TEXT ---
        # Equating the volumes
        equate_logic = VGroup(
            Text("आयतन बराबर हुन्छ:", font=nep_font, font_size=22),
            MathTex(r"V_{Cylinder} = V_{Total}"),
            MathTex(r"\pi \times 7^2 \times H = 1950.67"),
            MathTex(r"H \approx 12.67 \text{ cm}", color=YELLOW)
        ).arrange(DOWN).next_to(math_vgroup, DOWN, buff=0.5).to_edge(RIGHT)

        self.add_fixed_in_frame_mobjects(equate_logic)
        for line in equate_logic:
            self.play(Write(line))
            self.wait(0.5)

        self.play(Indicate(equate_logic[-1]), run_time=2)
        self.wait(2)


%manim -ql -v warning NepaliMathProblem
```
